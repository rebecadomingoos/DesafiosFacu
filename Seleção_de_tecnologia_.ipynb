```python
print('='*50)
print('Seleção para o projeto de tecnologia ')
print('='*50)

#..............................................
requisitos = {
       'python',
       'banco de dados',
       'git',
       'lógica',
       'html',
}

turnos_disponiveis = ('manhã', 'tarde')

#..............................................
nome= input('Digite seu nome: ')
idade= int(input('Digite sua idade: '))
curso= input('Digite seu curso: ')
semestre= int(input('Digite seu semestre: '))
email= input('Digite seu email: ')

#...............................................

conhecimentos_str = input(

    "Informe seus conhecimentos separados por vírgula: "
)
turnos_str = input(
    "Informe seu turno disponível separados por vírgula (manhã, tarde): "
  )

trabalha_em_equipe = input(
    "Aceita trabalhar em equipe? (sim/não) "
).strip().lower() == "sim"
computador_pessoal = input(
    "Possui computador próprio? (sim/não) "
).strip().lower() == "sim"


#............................................
conhecimentos_candidato = {
    c.strip().lower() for c in conhecimentos_str.split(",") if c.strip() != ""
}
turnos_candidato = {
    t.strip().lower() for t in turnos_str.split(",") if t.strip() != ""
}


codigo_inscricao = f"{nome[:3].upper()}-{curso[:3].upper()}-{semestre}"

#..............................................

candidato = {
    'nome': nome,
    'idade': idade,
    'curso': curso,
    'semestre': semestre,
    'email': email,
    'requisitos': conhecimentos_candidato,
    'turno': turnos_candidato,
    'trabalha_em_equipe': trabalha_em_equipe,
    'computador_pessoal': computador_pessoal,
    "codigo_inscricao": codigo_inscricao
}

#..............................................
conhecimentos_compativeis = candidato["requisitos"].intersection(
    requisitos,
)
conhecimentos_faltantes = requisitos.difference(
    candidato["requisitos"]
)
qtds_compativeis = len(conhecimentos_compativeis)

turno_validos = candidato["turno"].intersection(set(turnos_disponiveis))


email_valido = "@" in candidato["email"] and "." in candidato["email"]


#..............................................

idade_valida = candidato["idade"] >= 16
conhecimentos_validos = qtds_compativeis >= 3
turno_valido = len(turno_validos) > 0
aceita_equipe = candidato["trabalha_em_equipe"]

if turno_valido:
    pontuacao = len(conhecimentos_compativeis) * 2
    if "manhã" in turno_validos:
        pontuacao += 1
    if "tarde" in turno_validos:
        pontuacao += 1
    if candidato["trabalha_em_equipe"]:
        pontuacao += 2
    if candidato["computador_pessoal"]:
        pontuacao += 1
else:
    pontuacao = 0

aprovado = (
    idade_valida
    and conhecimentos_validos
    and turno_valido
    and aceita_equipe
    and pontuacao >= 7
)

if aprovado:
    classificacao = "APROVADO"
elif pontuacao >= 5:
    classificacao = "BANCO DE TALENTOS"
else:
    classificacao = "NÃO APROVADO"


destaque = pontuacao >= 12 and aprovado

#..............................................

motivos = []
if idade_valida == False:
    motivos.append("Idade mínima de 16 anos não atendida.")
if conhecimentos_validos == False:
    motivos.append("Menos de 3 conhecimentos compatíveis.")
if turno_valido == False:
    motivos.append(f"Nenhum turno compatível. Opções: {turnos_disponiveis}.")
if aceita_equipe == False:
    motivos.append("Não aceitou trabalhar em equipe.")
if email_valido == False:
    motivos.append("E-mail inválido (precisa conter '@' e '.').")
if pontuacao < 7:
    motivos.append("Pontuação final abaixo do mínimo para aprovação (7 pontos).")

#..............................................
print()
print("=" * 50)
print(" RESULTADO DA SELEÇÃO")
print("=" * 50)

#..............................................

print(f"Código de inscrição: {candidato['codigo_inscricao']}")
print(f"Nome: {candidato['nome']}")
print(f"Idade: {candidato['idade']}")
print(f"Curso: {candidato['curso']} - {candidato['semestre']}º semestre")
print(f"E-mail: {candidato['email']} ({'válido' if email_valido else 'inválido'})")
print(f"Turno informado: {candidato['turno']} (compatível: {turno_validos if turno_validos else 'nenhum'})")
print(f"Conhecimentos compatíveis ({qtds_compativeis}): {conhecimentos_compativeis if conhecimentos_compativeis else 'nenhum'}")
print(f"Conhecimentos faltantes: {conhecimentos_faltantes if conhecimentos_faltantes else 'nenhum'}")
print(f"Aceita trabalho em equipe: {'sim' if candidato['trabalha_em_equipe'] else 'não'}")
print(f"Possui computador próprio: {'sim' if candidato['computador_pessoal'] else 'não'}")
print(f"Pontuação final: {pontuacao} pontos")
print("-" * 45)
print(f"Classificação final: {classificacao}")

#..............................................
if destaque:
    print('Selo especial: "Candidato Destaque" ')
if motivos:
    print("\nMotivos que impactaram o resultado:")
    for motivo in motivos:
        print(f"  - {motivo}")
#..............................................
print("=" * 45)


```
